```python
from google.colab import drive
import cv2
import os
import matplotlib.pyplot as plt

# 1. Mount (Hubungkan) Google Drive
drive.mount('/content/drive')

# 2. Tentukan Path/Alamat File di Google Drive
file1_path = '/content/drive/MyDrive/PCVK/DATASET_JS_1/ktm1.jpeg'
file2_path = '/content/drive/MyDrive/PCVK/DATASET_JS_1/ktm2.jpg'

# Membaca gambar menggunakan OpenCV dan mengubah pewarnaan BGR ke RGB
img1 = cv2.cvtColor(cv2.imread(file1_path), cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(cv2.imread(file2_path), cv2.COLOR_BGR2RGB)

# 3. Tampilkan Kedua Citra (Dengan Sumbu/Garis Dimensi)
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Plot Gambar 1
axes[0].imshow(img1)
axes[0].set_title('KTM1.jpg (Original)')
# axes[0].axis('off')
# Plot Gambar 2
axes[1].imshow(img2)
axes[1].set_title('KTM2.jpg (Cropped)')
# axes[1].axis('off')

plt.tight_layout()
plt.show()

# 4. Tampilkan Dimensi Gambar & Ukuran File di Console
dim1 = img1.shape
dim2 = img2.shape
size1_kb = os.path.getsize(file1_path) / 1024
size2_kb = os.path.getsize(file2_path) / 1024

print("--- INFORMASI CITRA ---")
print(f"KTM1.jpg -> Dimensi: {dim1[1]}x{dim1[0]} px | Ukuran File: {size1_kb:.2f} KB")
print(f"KTM2.jpg -> Dimensi: {dim2[1]}x{dim2[0]} px | Ukuran File: {size2_kb:.2f} KB")
```
